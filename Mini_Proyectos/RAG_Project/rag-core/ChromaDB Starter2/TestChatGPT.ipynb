## Tests

### Test history chats

```python
client = OpenAI()
```

```python
## First communication

history = [
    {
        "role": "user",
        "content": "tell me a joke"
    }
]

response = client.responses.create(
    model="gpt-4o-mini",
    input=history,
    store=False
)

print(response.output_text)
```

```python
response
```

```python
# Add the response to the conversation
history += [{"role": el.role, "content": el.content} for el in response.output]

history.append({ "role": "user", "content": "tell me another" })

second_response = client.responses.create(
    model="gpt-4o-mini",
    input=history,
    store=False
)
```

```python
print(second_response.output_text)
```

### Test conversation

```python
conv_response = client.responses.create(
    model="gpt-4o-mini",
    input="tell me a joke",
)
print(conv_response.output_text)
```

```python
conv_second_response = client.responses.create(
    model="gpt-4o-mini",
    previous_response_id=conv_response.id,
    input=[{"role": "user", "content": "explain why this is funny."}],
)
print(conv_second_response.output_text)
```

```python
conv_third_response = client.responses.create(
    model="gpt-4o-mini",
    previous_response_id=conv_second_response.id,
    input=[{"role": "user", "content": "What was the first input I send?"}],
)
print(conv_third_response.output_text)
```

```python
conv_fourth_response = client.responses.create(
    model="gpt-4o-mini",
    previous_response_id=conv_third_response.id,
    input=[{"role": "user", "content": "Then, I just need to add the last ResponseID to chain the entire conversation?"}],
)
print(conv_fourth_response.output_text)
```
