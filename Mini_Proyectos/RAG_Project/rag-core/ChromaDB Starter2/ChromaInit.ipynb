### Initializing

```python
collection = chroma_client.get_or_create_collection(name="my_collection")
# get_or_create_collection
```

```python
collection.add(
    ids=["id1", "id2"],
    metadatas=[{"source": "OneSource"}, {"source": "AnotherSource"}],
    documents=[
        "This is a document about pineapple",
        "This is a document about oranges"
    ]
)
```

```python
collection.add(
    ids=["id40", "id41"],
    metadatas=[{"source": "OneSource", "page":1}, {"source": "AnotherSource", "page":14}],
    documents=[
        "This is a document about pineapple",
        "This is a document about oranges"
    ]
)
```

### Querying Chroma

```python
results = collection.query(
    query_texts=["This is a query document about hawaii"], # Chroma will embed this for you
    n_results=2 # how many results to return, default 10
)

results
```

## Menu Items

### Read file

```python
with open('files/menu_items.csv') as file:
    lines = csv.reader(file)

    documents = []
    metadatas = []
    ids = []
    id = 1

    for i, line in enumerate(lines):
        if (i == 0):
            continue

        documents.append(line[1])
        metadatas.append({"item_id": line[0]})
        ids.append(str(id))

        id+=1

len(documents)
```

```python
sentence_transformer_ef = embedding_functions.SentenceTransformerEmbeddingFunction(model_name="all-mpnet-base-v2")
# https://sbert.net/docs/sentence_transformer/pretrained_models.html
```

```python
collection_menu = chroma_client.get_or_create_collection(name="menu_collection", embedding_function=sentence_transformer_ef)
```

```python
collection_menu.add(
    ids = ids,
    metadatas = metadatas,
    documents = documents
)
```

```python
menu_results = collection_menu.query(
    query_texts=["shrimp"],
    n_results=5,
    include=["distances", "metadatas", "documents"]
)

menu_results
```
